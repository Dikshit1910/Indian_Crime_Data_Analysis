```python
plt.figure(figsize=(15,10))
df['City'].value_counts().plot.bar()
ax = plt.gca()
for container in ax.containers:
    ax.bar_label(container, label_type='center',color='white' , rotation=90)
plt.xlabel('Cities')
plt.ylabel('Values')
plt.title('Number of crimes Per City')
plt.show()
```

```python
vg = df['Victim Gender'].value_counts()
vg
```

```python
plt.figure(figsize=(10,3))
vg.plot.bar()
ax = plt.gca()
for container in ax.containers:
    ax.bar_label(container, label_type='center', color='white', fontweight='bold')

plt.title('Number of Crime by Gender')
plt.xlabel('GENDER')
plt.ylabel('Counts')
plt.show()
```

```python
va_top5 =df['Victim Age'].value_counts().head()
va_bottom5 =df['Victim Age'].value_counts().tail()

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# Plot top 5 on the first chart (axes[0])
va_top5.plot.bar(ax=axes[0], color='green')
axes[0].set_title("Top 5 Victim Ages")
axes[0].set_ylabel("Counts")
axes[0]= plt.gca()
for container in axes[0].containers:
    axes[0].bar_label(container, label_type='center', color='white', fontweight='bold')
    
# Plot bottom 5 on the second chart (axes[1])
va_bottom5.plot.bar(ax=axes[1], color='red')
axes[1].set_title("Bottom 5 Victim Ages")
axes[1] = plt.gca()
for container1 in axes[1].containers:
    axes[1].bar_label(container1, label_type='center', color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```

Highest Crimes happens with 25 years age person
